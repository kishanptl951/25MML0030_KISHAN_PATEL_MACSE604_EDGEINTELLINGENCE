```python
def take_photo(filename='photo.jpg', quality=0.8):
  js = Javascript('''
    async function takePhoto(quality) {
      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = 'Capture';
      div.appendChild(capture);

      const video = document.createElement('video');
      video.style.display = 'block';
      const stream = await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);
      video.srcObject = stream;
      await video.play();

      google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

      await new Promise((resolve) => capture.onclick = resolve);

      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      canvas.getContext('2d').drawImage(video, 0, 0);

      stream.getVideoTracks()[0].stop();
      div.remove();

      return canvas.toDataURL('image/jpeg', quality);
    }
  ''')
  display(js)
  data = eval_js('takePhoto({})'.format(quality))
  binary = b64decode(data.split(',')[1])

  with open(filename, 'wb') as f:
    f.write(binary)

  return filename
```

```python
 import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def extract_grayscale_frame(video_path, frame_number=0):
    # 1. Load the video
    cap = cv2.VideoCapture(video_path)

    # Set the frame position
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_number)
    success, frame = cap.read()

    if not success:
        print("Error: Could not read frame.")
        return None

    # 2. Convert to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 3. Calculate Metrics
    mean_val = np.mean(gray_frame)
    std_val = np.std(gray_frame)

    print(f"Metrics for Frame {frame_number}:")
    print(f"- Mean Intensity: {mean_val:.2f}")
    print(f"- Standard Deviation: {std_val:.2f}")

    # Display result in Colab
    cv2_imshow(gray_frame)

    cap.release()
    return gray_frame

# Usage
path = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
gray_img = extract_grayscale_frame(path, frame_number=0)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def process_video_frame(video_path, frame_number=0):
    cap = cv2.VideoCapture(video_path)
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_number)
    success, frame = cap.read()
    cap.release()

    if not success:
        return "Error: Could not read video."

    # Convert to Grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 1. Show Image Metrics
    print(f"--- Frame {frame_number} Metrics ---")
    print(f"Dimensions: {gray_frame.shape}")
    print(f"Mean Intensity: {np.mean(gray_frame):.2f}")
    print(f"Min/Max: {np.min(gray_frame)} / {np.max(gray_frame)}")

    # 2. Show the Matrix (Top-left 10x10 slice)
    print("\n--- Pixel Matrix (10x10 Slice) ---")
    print(gray_frame[:10, :10]) # Rows 0-9, Columns 0-9

    # 3. Show the Visual
    print("\n--- Visual Output ---")
    cv2_imshow(gray_frame)

# Usage
path = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
process_video_frame(path)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def divide_and_edge_detect(video_path, sample_rate=30):
    cap = cv2.VideoCapture(video_path)
    frame_count = 0

    if not cap.isOpened():
        print("Error: Video not found.")
        return

    print(f"--- Processing Video: {video_path} ---")

    while True:
        success, frame = cap.read()
        if not success:
            break  # Stop when video ends

        # 1. Grayscale Conversion
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # 2. Edge Detection (Canny)
        # 100 and 200 are the threshold values for edge intensity
        edges = cv2.Canny(gray, 100, 200)

        # 3. Display every 'sample_rate' frame to avoid crashing the browser
        if frame_count % sample_rate == 0:
            print(f"\n--- Frame {frame_count} Edge Map ---")
            print(f"Edge Pixel Matrix (Top 5x5):\n{edges[:5, :5]}")
            cv2_imshow(edges)

        frame_count += 1

    cap.release()
    print(f"\nTotal frames processed: {frame_count}")

# Usage: sample_rate=30 shows 1 frame per second (assuming 30fps video)
path = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
divide_and_edge_detect(path, sample_rate=50)
```

```python
import cv2
import os

def detect_video_edges(input_path, output_path, low_threshold=50, high_threshold=150):
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return

    # Get properties for the output file
    width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps    = cap.get(cv2.CAP_PROP_FPS)

    # Use mp4v codec for Colab compatibility
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height), isColor=False)

    print("Detecting edges... This may take a moment.")

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Step 1: Grayscale conversion (Edge detection works on intensity)
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # Step 2: Gaussian Blur to remove noise
        blurred = cv2.GaussianBlur(gray, (5, 5), 0)

        # Step 3: Canny Edge Detection
        edges = cv2.Canny(blurred, low_threshold, high_threshold)

        out.write(edges)

    cap.release()
    out.release()
    print(f"Edge detection complete! Saved to: {output_path}")

# Run the detection
input_file = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
edge_output = '/content/drive/MyDrive/Edge_Intelligence/edge_detected_video.mp4'
detect_video_edges(input_file, edge_output)
```

```python
import cv2
import os

def convert_video_to_grayscale(input_path, output_path):
    # 1. Open the video file
    cap = cv2.VideoCapture(input_path)

    if not cap.isOpened():
        print("Error: Could not open video.")
        return

    # 2. Get original video properties (width, height, fps)
    width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps    = cap.get(cv2.CAP_PROP_FPS)

    # 3. Define the codec and create VideoWriter object
    # 'mp4v' works well for .mp4 files in Colab
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')

    # IMPORTANT: isColor=False tells the writer to expect grayscale frames
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height), isColor=False)

    print("Processing video... Please wait.")

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # 4. Convert current frame to grayscale
        gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # 5. Write the grayscale frame to the output file
        out.write(gray_frame)

    # 6. Release everything when finished
    cap.release()
    out.release()
    print(f"Success! Grayscale video saved at: {output_path}")

# --- EXECUTION ---
input_file = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
output_file = '/content/drive/MyDrive/Edge_Intelligence/grayscale_demo_video.mp4'

convert_video_to_grayscale(input_file, output_file)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow # Special function for displaying images in Google Colab
import time

# Define the path to your video file
# You may need to upload a file to your Colab environment or mount Google Drive
video_path = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'

# Create a VideoCapture object to read the video from the file path
cap = cv2.VideoCapture(video_path)

# Check if the video file was opened successfully
if not cap.isOpened():
    print("Error: Could not open video file.")
else:
    print("Video file opened successfully. Starting processing...")

# Loop through the video frames
while cap.isOpened():
    # Read a frame from the video
    ret, frame = cap.read()

    # If the frame was read correctly, ret is True
    if ret:
        # Convert the color frame to grayscale using cv2.cvtColor()
        gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # *** FIND THE NEGATIVE OF THE IMAGE ***
        # Method 1: Using cv2.bitwise_not() (recommended for simplicity)
        negative_frame = cv2.bitwise_not(gray_frame) #

        # Method 2: Using array subtraction (as images are NumPy arrays)
        # negative_frame = 255 - gray_frame #

        # Display the negative frame in Google Colab
        cv2_imshow(negative_frame)

        # Optional: Add a small delay to control playback speed
        # Adjust the sleep time (in seconds) to achieve the desired frame rate
        time.sleep(0.03)
    else:
        # Break the loop if the video ends or an error occurs
        print("End of video or error reading frame.")
        break

# Release the video capture object and close any open windows
cap.release()
# cv2.destroyAllWindows() # Not necessary for cv2_imshow in Colab
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import time

video_path = '/content/drive/MyDrive/Edge_Intelligence/real_world_demo_video.mp4'
cap = cv2.VideoCapture(video_path)

# Initialize the first frame for comparison
ret, prev_frame = cap.read()
if ret:
    prev_gray = cv2.cvtColor(prev_frame, cv2.COLOR_BGR2GRAY)
    prev_gray = cv2.GaussianBlur(prev_gray, (21, 21), 0) # Smooth to reduce noise

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 1. Pre-process current frame
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (21, 21), 0)

    # 2. Calculate the Difference (Motion)
    # This subtracts the previous frame from the current one
    frame_delta = cv2.absdiff(prev_gray, gray)

    # 3. Thresholding
    # If the change is > 25, turn pixel white (motion), else black (still)
    thresh = cv2.threshold(frame_delta, 25, 255, cv2.THRESH_BINARY)[1]

    # 4. Dilate the threshold to fill in holes
    thresh = cv2.dilate(thresh, None, iterations=2)

    # Display the result
    print("--- Motion Mask ---")
    cv2_imshow(thresh)

    # Update the previous frame for the next iteration
    prev_gray = gray

    # Control playback speed
    time.sleep(0.01)

cap.release()
```
